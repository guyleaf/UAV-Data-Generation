```python
import numpy as np
import pandas as pd
import glob
import os
```

```python
path = os.path.expanduser("~/data/Background/Unsplash_1_13_0/original")
documents = ["photos", "keywords", "collections", "conversions", "colors"]
datasets = {}

for doc in documents:
    files = sorted(glob.iglob(os.path.join(path, f"{doc}.csv*")))

    subsets = []
    for filename in files:
        df = pd.read_csv(filename, sep="\t", header=0)
        subsets.append(df)

    datasets[doc] = pd.concat(subsets, axis=0, ignore_index=True)
```
